`x or y` возвращает `x`, если `x` истинно, иначе вычисляет и возвращает `y`.
Это выбор **операнда**, а не обязательное преобразование ответа в `bool`.
Число `0` ложно в логическом контексте, поэтому `reading or "нет данных"`
возвращает строку `"нет данных"`. При `reading = 5` вернулось бы число `5`.

Однако нулевое измерение может быть совершенно корректным. Если отсутствие данных
обозначено именно `None`, проверяем `reading is not None`: второе выражение
сохраняет `0`. Ложными также считаются `False`, пустые строки и коллекции;
`or` не различает эти значения и отсутствие данных.

### Оператор `==` vs `is`

````python
# Равенство значений не означает идентичность объектов
measurements = [18, 21]
alias = measurements
independent = measurements.copy()
measurements == independent, measurements is independent, measurements is alias
````

````python
alias.append(23)
measurements, independent
````

Присваивание связывает имя с объектом и не копирует данные.
`==` сравнивает значения, `is` — идентичность объектов. Для отсутствующего значения
пишут `value is None`.

**Интернирование строк** — переиспользование одного объекта для одинаковых строк.
Python может делать это автоматически; явно интернировать строку позволяет
`sys.intern`. В CPython также есть кэш некоторых целых чисел — близкая оптимизация
переиспользования объектов. Это не означает, что любые равные числа или строки
обязательно являются одним объектом. Поэтому содержимое сравнивают через `==`,
а не через `is`: результат не должен зависеть от подобных оптимизаций.

````python
# Поверхностная копия сохраняет ссылки на вложенные объекты
rows = [[18, 21], [19, 22]]
rows_copy = rows.copy()
rows_copy[0].append(23)
rows, rows_copy
````

### Поверхностная и глубокая копии

`list.copy()` создаёт новый внешний список, но его элементы — ссылки на те же
объекты. В примере выше изменение вложенного списка видно в обеих копиях.
`deepcopy` из стандартного модуля `copy` рекурсивно копирует вложенные объекты:
для списка списков получаются независимые вложенные списки.

````python
from copy import deepcopy

original_rows = [[18, 21], [19, 22]]
shallow_rows = original_rows.copy()
deep_rows = deepcopy(original_rows)

original_rows is shallow_rows, original_rows[0] is shallow_rows[0], original_rows[0] is deep_rows[0]
````

````python
# Поверхностная копия разделяет вложенный список с оригиналом
shallow_rows[0].append(23)

# Глубокая копия содержит собственный вложенный список
deep_rows[1].append(99)

print("Оригинал:", original_rows)
print("Поверхностная копия:", shallow_rows)
print("Глубокая копия:", deep_rows)
````

Добавление элемента во **внешний** список поверхностной копии не изменит внешний
список оригинала: они уже разные. Общими остаются именно вложенные объекты.
`deepcopy` обычно требует больше работы и памяти; она нужна, когда изменения
вложенной структуры должны быть независимыми.

#### defaultdict: создание значения при отсутствии ключа

`defaultdict` получает **фабрику значений** — функцию без аргументов.
При обращении `dictionary[key]` по отсутствующему ключу она вызывается,
а её результат сохраняется в словаре под этим ключом. Например, `list`
создаёт новый пустой список для каждой новой группы.

````python
from collections import defaultdict

observations = [("north", 18), ("south", 23), ("north", 21)]
by_station = defaultdict(list)  # передаём list, а не результат вызова list()

for station, temperature in observations:
    by_station[station].append(temperature)

dict(by_station)
````

````python
# get не вызывает фабрику и не добавляет ключ даже у defaultdict
print(by_station.get("west"))
print("west" in by_station)

# Обращение через [] создаёт и сохраняет пустой список
print(by_station["west"])
print("west" in by_station)
````

Операция, похожая на чтение `by_station["west"]`, здесь меняет состояние словаря.
Это удобно для накопления и группировки данных, но важно учитывать при проверках
наличия ключа. `key in dictionary` и `dictionary.get(key)` записей не создают.

`def` связывает имя с функцией. Её можно вызвать, передать в `sorted(key=...)`,
`map` или другую функцию. `lambda` удобна для короткого выражения;
имя и `def` обычно яснее для нетривиального преобразования.

````python
# Именованные функции
def power(value, degree=2):
    return value**degree


# Анонимная функция
apower = lambda value, degree=2: value**degree

print(power(2))
print(apower(2))
````

````python
print(power(2, 3))
print(apower(2, 3))
````

### Вход, выход и отсутствие скрытой мутации

Чистое преобразование получает все необходимые данные аргументами,
возвращает результат и не меняет внешнее состояние. Это упрощает повторный
запуск и проверку: одинаковому входу соответствует одинаковый результат.

````python
def normalize_scores(scores, maximum=100):
    return [score / maximum for score in scores]

scores = [60, 75, 90]
snapshot = scores.copy()
normalized = normalize_scores(scores)
assert scores == snapshot
assert normalized == normalize_scores(scores)
normalized
````

### Внутренние функции · справочник

````python
def mse(true_values, pred_values):
    def squared_error(true, predicted):
        return (true - predicted) ** 2

    if not true_values:
        raise ValueError("Нужно хотя бы одно наблюдение")
    pairs = zip(true_values, pred_values, strict=True)
    return sum(squared_error(true, pred) for true, pred in pairs) / len(true_values)
````

````python
mse([1,2,3], [1.2,1.8,3.1])
````

````python
def logger(function):
    def _logger(*args, **kwargs):
        print("started: {}".format(function.__name__))
        result = function(*args, **kwargs)
        print("finished: {}".format(function.__name__))
        return result
    return _logger
````

````python
@logger
def do_something(activity="it"):
    print("doing {}!".format(activity))
    return True

@logger
def make_something(activity="it"):
    print("making {}!".format(activity))
    return True
````

### Ошибка: последний индекс не равен правой границе среза

Сокращённый фрагмент прежнего ручного разбиения слов. После цикла `i` указывает
на последний символ, но правая граница среза в результат не входит.
Для пустой строки цикл вообще не связывает имя `i`.

````python
def last_word_wrong(text):
    start_index = 0
    for i in range(len(text)):
        if text[i] == " ":
            start_index = i + 1
    return text[start_index:i]  # ошибка: последний символ потерян

last_word_wrong("слово"), last_word_wrong("два слова")
````

````python
try:
    last_word_wrong("")
except UnboundLocalError as error:
    print(f"Ожидаемая ошибка: {error}")
````

Для хвоста нужна граница `text[start_index:]`, а начало должно быть определено
и для пустого ввода. Но здесь ручной проход не нужен: `split()` уже обрабатывает
пустую строку, повторяющиеся пробелы и табуляцию. Ниже сравним список и генератор
с одинаковым корректным разбиением.

````python
def create_word_list(text):
    """Слова без служебного маркера bot; split учитывает пробельные символы."""
    return [word for word in text.split() if word != "bot"]
````

````python
TEXT = "  данные bot и\tфункции "
````

````python
create_word_list(TEXT)
````

````python
def create_word_generator(text):
    """Тот же отбор, но подходящие слова выдаются по запросу."""
    for word in text.split():
        if word != "bot":
            yield word
````

````python
create_word_generator(TEXT)
````

````python
gen = create_word_generator(TEXT)
next(gen), list(gen), list(gen)
````

````python
# Использование в цикле
for word in create_word_generator(TEXT):
    print(word)
````

````python
# Преобразование в список
list(create_word_generator(TEXT))
````

Генератор — одноразовый итератор: уже прочитанные элементы не возвращаются.
`yield` откладывает выдачу элементов, но `text.split()` в этом примере всё равно
создаёт список слов целиком. Ленивость одного этапа не делает всю обработку потоковой.

````python
for text in ["", "слово", "bot", "данные bot функции"]:
    assert create_word_list(text) == list(create_word_generator(text))
assert create_word_list("слово") == ["слово"]
````

## Классы

````python
class Person:
    """Объект хранит состояние; метод использует его при вызове."""

    def __init__(self, first_name, second_name):
        self.first_name = first_name
        self.second_name = second_name

    def full_name(self):
        return f"{self.first_name} {self.second_name}"
````

````python
alex = Person("Алексей", "Смирнов")
alex.full_name()
````

````python
alex.first_name = "Александр"
alex.full_name()
````

Объект объединяет данные и поведение. Ответ `full_name()` зависит от текущего
состояния `alex`. Это полезная организация данных, но изменение атрибутов
нужно учитывать при повторных вызовах. Свойства, наследование и специальные
методы подробно рассматриваются на занятии о классах.

### Подключение собственных модулей

Модуль — файл `.py` с определениями. Ниже создаётся временный `custommath.py`,
который можно импортировать. В обычном проекте такой файл сохраняют рядом с кодом.
Проверка `__name__ == "__main__"` отделяет запуск файла от его импорта.

````python
import sys
from importlib import import_module

module_source = '''
def power(value, degree):
    """Возводит значение в степень."""
    return value ** degree

if __name__ == "__main__":
    print("2 ** 3 =", power(2, 3))
'''

with TemporaryDirectory() as directory:
    module_path = Path(directory) / "custommath.py"
    module_path.write_text(module_source, encoding="utf-8")
    sys.path.insert(0, directory)
    try:
        custommath = import_module("custommath")
        print(custommath.power(2, 3))
    finally:
        sys.path.remove(directory)
        sys.modules.pop("custommath", None)
````

Импорт выполняет верхний уровень модуля один раз и сохраняет модуль в
`sys.modules`. В демонстрации путь и запись в этом кэше убраны после использования.
При изменении собственного модуля в Jupyter перезапустите ядро перед проверкой:
старый импорт не обновляет код автоматически.
